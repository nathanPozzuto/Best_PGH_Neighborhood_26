# Introduction

I chose to do the police arrests in the city of Pittsburgh from 2016 to 2025. This dataset will allow me to determine which neighborhood has the most arrests occuring in them, which showcases the city with the most responding police units and the most dangerous people (in theory this is what it means, but it may not be related). 


# Sub-Metric

The sub-metric we chose to use was the 'safest' neighborhood. For me to determine this, I will be looking at the neighborhood with the least amount of arrests in the time frame of the data, as well as possibly the neighborhood with the most arrests. showcasing the most responsive neighborhood.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

CSV_PATH = "arrests.csv"   # <-- change to your file path
TOP_N = 10

# Load & clean
df = pd.read_csv(CSV_PATH, encoding="utf-8-sig")
df["ARRESTTIME"] = pd.to_datetime(df["ARRESTTIME"], errors="coerce")
df["year"] = df["ARRESTTIME"].dt.year
df = df.dropna(subset=["INCIDENTNEIGHBORHOOD", "year"])
df["year"] = df["year"].astype(int)
df = df[df["year"] >= 2016]  # a few stray pre-2016 records; 2016 starts in August

# Aggregate
counts = df.groupby(["INCIDENTNEIGHBORHOOD", "year"]).size().unstack(fill_value=0)
top = counts.sum(axis=1).nlargest(TOP_N).index
top_counts = counts.loc[top]

# Plot
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(17, 7),
                               gridspec_kw={"width_ratios": [1.1, 1]})

# Left: heatmap
data = top_counts.values
im = ax1.imshow(data, cmap="YlOrRd", aspect="auto")
ax1.set_xticks(range(len(top_counts.columns)))
ax1.set_xticklabels(top_counts.columns)
ax1.set_yticks(range(len(top_counts.index)))
ax1.set_yticklabels(top_counts.index)
ax1.grid(False)
threshold = data.max() * 0.6
for i in range(data.shape[0]):
    for j in range(data.shape[1]):
        ax1.text(j, i, f"{data[i, j]}", ha="center", va="center",
                 color="white" if data[i, j] > threshold else "black", fontsize=9)
fig.colorbar(im, ax=ax1, label="Number of arrests")
ax1.set_title(f"Arrests per Year – Top {TOP_N} Neighborhoods", fontsize=14, weight="bold")
ax1.set_xlabel("Year")

# Right: line chart
colors = plt.cm.tab10(np.linspace(0, 1, 10))
for k, name in enumerate(top_counts.index):
    ax2.plot(top_counts.columns, top_counts.loc[name], marker="o",
             label=name, color=colors[k % 10])
ax2.set_title("Arrest Trends Over Time", fontsize=14, weight="bold")
ax2.set_xlabel("Year")
ax2.set_ylabel("Number of arrests")
ax2.set_xticks(list(top_counts.columns))
ax2.grid(True, alpha=0.3)
ax2.legend(title="Neighborhood", fontsize=9, loc="upper right")

fig.suptitle("Neighborhoods with the Highest Arrest Counts\n"
             "(2016 data begins in August, so that year is partial)",
             fontsize=12, y=1.02)
plt.tight_layout()
plt.savefig("arrests_by_neighborhood.png", dpi=150, bbox_inches="tight")
plt.show()
```
